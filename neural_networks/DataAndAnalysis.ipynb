#### Sample Model

```python
dims = X_train.shape[0]

w = np.random.randn(dims, 1)
b = np.random.randn()

assert(w.shape == (dims, 1))
assert(isinstance(b, float) or isinstance(b, int))

num_iterations = 2000

for i in range(num_iterations):
    
    Z = np.dot(w.T, X_train) + b
    assert(Z.shape == (w.shape[1], X_train.shape[1]))
    
    #A = np.maximum(0, Z)
    #assert(A.shape == Z.shape)
    
    m = X_train.shape[1]
    #cost = (1/(2 * float(m))) * np.sum((A - Y_train) ** 2) # Relu
    cost = (1/(2 * float(m))) * np.sum((Z - Y_train) ** 2) # Linear
    cost = np.squeeze(cost)
    assert(cost.shape == ())
    
    #dA = A - Y_train
    #assert(dA.shape == A.shape)
    
    #dZ = np.array(dA, copy = True) #Relu
    #dZ[Z <= 0] = 0 #Relu
    dZ = Z - Y_train
    assert(dZ.shape == Z.shape)
    
    dw = (1/float(m)) * np.dot(X_train, dZ.T)
    db = (1/float(m)) * np.sum(dZ)

    assert(dw.shape == w.shape)
    assert(db.dtype == float)
    
    learning_rate = 0.05

    w = w - learning_rate * dw
    b = b - learning_rate * db
    
    if i % 100 == 0:
        print ("Cost after iteration %i: %f" %(i, cost))
```
