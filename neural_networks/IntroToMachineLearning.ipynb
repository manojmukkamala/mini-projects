### Basic Data Exploration

```python
import numpy as np
import pandas as pd
```

```python
melbourne_file_path = "melbournehousingmarket/Melbourne_housing_FULL.csv"
```

```python
melbourne_data = pd.read_csv(melbourne_file_path)
```

### Machine Learning Model

```python
melbourne_data.columns
```

```python
melbourne_data = melbourne_data.dropna(axis = 0)
```

```python
y = melbourne_data["Price"]
```

```python
melbourne_features = ["Rooms", "Bathroom", "Landsize", "Lattitude", "Longtitude"]
```

```python
X = melbourne_data[melbourne_features]
```

```python
X.head()
```

```python
X.describe()
```

```python
from sklearn.tree import DecisionTreeRegressor

melbourne_model = DecisionTreeRegressor()
melbourne_model.fit(X, y)
```

```python
print("Making predictions for the following 5 houses:")
print(X.head())
print(y.head())
print("The predictions are")
print(melbourne_model.predict(X.head()))
```

### Model Validation

```python
from sklearn.metrics import mean_absolute_error

predicted_home_prices = melbourne_model.predict(X)
mean_absolute_error(y, predicted_home_prices)
```

```python
from sklearn.model_selection import train_test_split

train_X, val_X, train_y, val_y = train_test_split(X, y, random_state = 0)

melbourne_model = melbourne_model.fit(train_X, train_y)

val_predictions = melbourne_model.predict(val_X)

print(mean_absolute_error(val_y, val_predictions))
```

### Underfitting and Overfitting

- Overfitting: capturing spurious patterns that won't recur in the future, leading to less accurate predictions, or
- Underfitting: failing to capture relevant patterns, again leading to less accurate predictions.


We use validation data, which isn't used in model training, to measure a candidate model's accuracy. This lets us try many candidate models and keep the best one.

```python
def get_mae(max_leaf_nodes, train_X, train_y, val_X, val_y):
    melbourne_model = DecisionTreeRegressor(max_leaf_nodes = max_leaf_nodes, random_state = 0)
    melbourne_model.fit(train_X, train_y)
    preds_val = melbourne_model.predict(val_X)
    mae = mean_absolute_error(val_y, preds_val)
    return mae
```

```python
for max_leaf_nodes in [5, 10, 50, 100, 500, 1000, 5000]:
    my_mae = get_mae(max_leaf_nodes, train_X, train_y, val_X, val_y)
    print("Max leaf nodes: %d \t\t Mean Absolute Error: %d" %(max_leaf_nodes, my_mae))
```

### Random Forests

```python
from sklearn.ensemble import RandomForestRegressor

forest_model = RandomForestRegressor()
forest_model.fit(train_X, train_y)
melb_preds = forest_model.predict(val_X)
print(mean_absolute_error(val_y, melb_preds))
```
