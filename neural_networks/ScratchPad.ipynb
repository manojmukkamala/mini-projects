```python
tot_uniq_items = int(input("Number of distinct items available in the collection:"))
items = np.random.randint(1, tot_uniq_items+1, 1000000)
```

```python
uniq_items = []
num_of_boxes = []
box_count = 0

for cur_item in items:
    box_count += 1
    
    if cur_item not in uniq_items:
        uniq_items.append(cur_item)
        
    if len(uniq_items) == tot_uniq_items:
        num_of_boxes.append(box_count)
        box_count = 0
        uniq_items = []
```

```python
print("Average Number of boxes required to collect all the stickers:", np.mean(num_of_boxes))
print("When you are lucky, you can get all the 6 stickers in the first", np.min(num_of_boxes), "boxes")
print("But if are unlucky, you have to purchase", np.max(num_of_boxes), "boxes to get all the 6 stickers")
```

```python
plt.hist(num_of_boxes, ec = 'k')
plt.show()
```

### Central Limit Theorem with a Crazy Distribution

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.random.randint(1, 10, 100000)
```

```python
x = [i for i in x if i not in [4, 7, 8]]
```

```python
plt.hist(x)
plt.show()
```

```python
n = 10
s = np.random.randint(0, len(x), 100000)
```

```python
p = []
for i in range(0, len(s), n):
    q = []
    for j in s[i:i+n]:
        q.append(x[j])
    p.append(np.mean(q))
```

```python
plt.hist(p, ec = 'k')
plt.show()
```

```python
print("Population Mean:", np.mean(x))
print("Mean of Sampling Distribution of Sample Means:", np.mean(p))
```

```python
print("Population Standard Deviation:", np.std(x))
print("StdDev of Sampling Distribution of Sample Means:", np.std(x)/np.sqrt(n))
print("StdDev of Sampling Distribution of Sample Means numpy:", np.std(p))
```

```python
print("Number of trials: ", trials)
print("Number of Experiments: ", len(binomial))
```

```python
print("Expected Value of binomial: ", trials * p)
print("Mean of binomial:", np.mean(binomial))
```

```python
print("Variance of binomial:", trials * p * (1-p))
print("Standard Deviation of binomial:", np.sqrt(trials * p * (1-p)))
```

### Bernouli and Binomial Distributions (with large Sample)

```python
Y = np.random.randint(0, 2, 100000)
plt.hist(Y)
plt.show()
```

```python
p = sum(Y == 1)/len(Y)
print("E(Y): ", p)
print("Var(Y): ", p * (1-p))
print("Mean(Y) numpy: ", np.mean(p))
print("Var(Y) numpy:", np.var(Y))
```

```python
X = []
trials = 100
n = trials * 1000
for i in range(0, n, trials):
    X.append(sum(Y[i:(i+trials)]))
```

```python
plt.hist(X)
plt.show()
```

```python
print("E(X): ", trials * p)
print("Mean(X) numpy:", np.mean(X))
```

```python
print("Var(X): ", trials * p * (1-p))
print("Variance(X) numpy:", np.var(X))
```

```python
print("SD(X): ", np.sqrt(trials * p * (1-p)))
print("StdDev(X) numpy:", np.std(X))
```

```python
p_hat = np.divide(X, trials)
plt.hist(p_hat)
plt.show()
```

```python
print("Mean of p_hat:", p)
print("Mean of p_hat numpy:", np.mean(p_hat))
```

```python
print("StdDev of p_hat:", np.sqrt((p*(1-p))/trials))
print("StdDev of p_hat numpy:", np.std(p_hat))
```

```python
from abc import ABC, abstractclassmethod
class Shape(ABC):
    @abstractclassmethod
    def area(self): pass
    @abstractclassmethod
    def perimeter(self): pass
```

```python
class Square(Shape):
    def __init__(self, side):
        self.side = side
    def area(self): return self.side ** 2
    def perimeter(self): return 4 * self.side
```

```python
shape = Shape()
```

```python
shape.perimeter()
```

```python
square = Square(5)
```

```python
square.area()
```
